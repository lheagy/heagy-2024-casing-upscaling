```python
def plot_casing(model, xlim=None, zlim=None, vmin=None, vmax=None, ax=None, cb_label=None, logscale=True, show_cb=False): 
    if ax is None: 
        fig, ax = plt.subplots(1, 1)
    if logscale is True: 
        pcolor_opts={"norm":LogNorm(vmin=vmin, vmax=vmax), "cmap":"Spectral_r"}
    else: 
        pcolor_opts={"norm":Normalize(vmin=vmin, vmax=vmax), "cmap":"Spectral_r"}
    out = mesh.plot_image(
        model, pcolor_opts=pcolor_opts, ax=ax, mirror=True
    )

    if show_cb is True: 
        cb = plt.colorbar(out[-1], shrink=0.5)
        cb.set_label(cb_label)

    if xlim is not None: 
        ax.set_xlim(xlim)
    if zlim is not None: 
        ax.set_ylim(zlim)

    ax.set_xlabel("x (m)")
    ax.set_ylabel("z (m)")
    return ax, out
```

```python
def create_simulation(
    mesh, 
    sigma, mur, 
    rx_x, rx_z, 
    src_z, src_dz, 
    frequencies,
    components=["x", "z"], 
    inversion=False, true_inds=False
):
    rx_list = utils.create_receivers(rx_x, rx_z, components=components)
    src_list = []
    src_locations = np.array([
        [0, 0, src_z+src_dz/2.],
        [0, 0, src_z-src_dz/2.]
    ])
    for f in frequencies: 
        src = fdem.sources.LineCurrent(rx_list, location=src_locations, frequency=f)
        src_list.append(src)

    if inversion is True:
        wires = maps.Wires(("sigma", 1), ("mur", 1))
        sigma_map, permeability_map = utils.create_maps(
            mesh, sigma, mur, 
            casing_a, casing_b, casing_t, casing_l, 
            true_inds=true_inds
        )
    else: 
        wires = maps.Wires(("sigma", mesh.n_cells), ("mur", mesh.n_cells))
        permeability_map = maps.MuRelative(mesh)
        sigma_map = maps.ExpMap(mesh) 
    
    survey = fdem.Survey(src_list)
    
    sim = fdem.simulation.Simulation3DMagneticField(
        mesh=mesh, sigmaMap=sigma_map * wires.sigma, muMap=permeability_map * wires.mur,
        survey=survey, solver=get_default_solver(), forward_only=True
    )
    return sim
```

```python
def run_simulation(key, return_fields=False):
    sim = create_simulation(
        mesh, 
        models[key]["sigma"], models[key]["mur"], 
        rx_x, rx_z,
        src_z, src_dz, 
        frequencies=frequencies,
        components=["x", "z"],  
        inversion=False, true_inds=False
    )
    filename = os.path.sep.join([directory_path, key])
    
    if os.path.exists(f"{filename}.npy"): 
        print(f"loading {key}...")
        f = np.load(f"{filename}.npy")
        sim.model = np.hstack([np.log(models[key]["sigma"]), models[key]["mur"],])
        fields = sim.fieldsPair(sim)
        fields[:, "hSolution"] = f
    else: 
        t = time.time()
        print(f"starting {key}..., saving to {filename}")
        fields = sim.fields(np.r_[np.log(models[key]["sigma"]), models[key]["mur"]])
        print(f" ... done. {time.time()-t:1.2e} \n")
        np.save(filename, fields[:, "hSolution"])
    
    dobs = sim.make_synthetic_data(
        np.r_[models[key]["sigma"], models[key]["mur"]],
        f=fields
    )
    if return_fields:
        return sim, fields, dobs
    return sim, dobs
```

```python
def plot_fields(key=keys[0], freq_ind=0):
    fig, ax = plt.subplots(1, 4, figsize=(18, 5))
    ax=ax.reshape((2, 2), order="F")
    
    xlim = 0.1*np.r_[-1, 1]
    zlim = np.r_[-550, 10]
    
    j_plot = (mesh.average_face_to_cell_vector * fields[key][fields[key].simulation.survey.source_list[freq_ind], "j"])
    h_plot = (mesh.average_edge_to_cell * fields[key][fields[key].simulation.survey.source_list[freq_ind], "b"])
     
    for i, component in enumerate(["real", "imag"]): 
        out = mesh.plot_image(
            getattr(j_plot, component), v_type="CCv", view="vec",
            range_x = xlim, range_y = zlim,
            sample_grid = np.r_[np.diff(xlim)/100, np.diff(zlim)/100],
            mirror=True,
            ax=ax[0, i], 
            pcolor_opts={"norm":LogNorm(vmin=1e-10)}
        )
        cb = plt.colorbar(out[0], ax=ax[0, i])
        # cb.set_label("J (A/m$^2$)")

        hplot_component = getattr(h_plot, component)
        vmax = np.max(np.abs(hplot_component))
        out = mesh.plot_image(
            hplot_component, 
            range_x = xlim, range_y = zlim,
            sample_grid = np.r_[np.diff(xlim)/100, np.diff(zlim)/100],
            mirror=True, mirror_data=-hplot_component,
            ax=ax[1, i], 
            pcolor_opts={"norm":SymLogNorm(linthresh=1e-9, vmin=-vmax, vmax=vmax), "cmap":"RdBu_r"}
        )
        cb = plt.colorbar(out[0], ax=ax[1, i])
        # cb.set_label("H (A/m)")
    
        ax[0, i].set_title(f"J, {component}")
        ax[1, i].set_title(f"B, {component}")
    
    t_over_delta = casing_t/skin_depth(models[key]['sigma'].max(), models[key]['mur'].max(), frequencies[freq_ind])
    fig.suptitle(f"{key}, {frequencies[freq_ind]:1.2e} Hz, t/$\delta$ = {t_over_delta:1.2e}")
    plt.tight_layout()
```

```python
def plot_fields_compare(key=keys[0], freq_ind=0, field="j", subtract=None, x_extent=0.1, depth=550, vmin=None):
    fig, ax = plt.subplots(2, 4, figsize=(18, 10))
    
    xlim = x_extent*np.r_[-1, 1]
    zlim = np.r_[-depth, 10]

    f_true = fields[key][fields[key].simulation.survey.source_list[freq_ind], field].squeeze()
    f_inv = fields_upscaled[key][frequencies[freq_ind]][:, field].squeeze()
    f_dc = fields_dc[key][fields_dc[key].simulation.survey.source_list[freq_ind], field].squeeze()
    f_high = fields_highf[key][fields_highf[key].simulation.survey.source_list[freq_ind], field].squeeze()

    if subtract is not None: 
        if subtract.lower() == "true": 
            f_subtract = f_true
            # f_true = f_true - f_subtract
            f_inv = f_inv - f_subtract
            f_dc = f_dc - f_subtract
            f_high = f_high - f_subtract
        elif subtract.lower() == "inv": 
            f_subtract = f_inv
            f_true = f_true - f_subtract
            # f_inv = f_inv - f_subtract
            f_dc = f_dc - f_subtract
            f_high = f_high - f_subtract
        elif subtract.lower() == "dc":
            f_subtract = f_inv
            f_true = f_true - f_subtract
            f_inv = f_inv - f_subtract
            # f_dc = f_dc - f_subtract
            f_high = f_high - f_subtract
        elif subtract.lower() == "high": 
            f_subtract = f_high
            f_true = f_true - f_subtract
            f_inv = f_inv - f_subtract
            f_dc = f_dc - f_subtract
            # f_high = f_high - f_subtract
        
    
    if field in ["e", "j"]:
        ave = mesh.average_face_to_cell_vector
        mirror_factor = 1
        v_type="CCv"
        view="vec"
        if vmin is None or vmin == 0: 
            vmin=1e-7*np.max(np.abs(f_true))
        pcolor_opts={"norm":LogNorm()}
    elif field in ["b", "h"]:
        ave = mesh.average_edge_to_cell
        mirror_factor = -1
        v_type="CC"
        view="real"
        vmax = np.max(np.abs(f_true))
        pcolor_opts={"norm":SymLogNorm(linthresh=1e-9, vmin=-vmax, vmax=vmax), "cmap":"RdBu_r"}
        
    for j, plotme in enumerate([f_true, f_inv, f_dc, f_high]):
        for i, component in enumerate(["real", "imag"]): 
            plot_data = ave * getattr(plotme, component)
            out = mesh.plot_image(
                plot_data, 
                v_type=v_type, view=view,
                range_x = xlim, range_y = zlim,
                sample_grid = np.r_[np.diff(xlim)/100, np.diff(zlim)/100],
                mirror=True,
                mirror_data=mirror_factor*plot_data,
                ax=ax[i, j], 
                pcolor_opts=pcolor_opts
            )
            cb = plt.colorbar(out[0], ax=ax[i, j])
    
    ax[0,0].set_title(f"True {sigma_casing:1.1e} S/m, ({field})")
    ax[0,1].set_title(f"Inverted {np.exp(logsig_min[key][freq_ind]):1.1e} S/m, ({field})")
    ax[0,2].set_title(f"DC {sigdc:1.1e} S/m, ({field})")
    ax[0,3].set_title(f"High Freq {sigma_casing:1.1e}, ({field})")    
    
    t_over_delta = casing_t/skin_depth(models[key]['sigma'].max(), models[key]['mur'].max(), frequencies[freq_ind])
    fig.suptitle(f"{key}, {frequencies[freq_ind]:1.2e} Hz, t/$\delta$ = {t_over_delta:1.2e}")
    plt.tight_layout()
```

```python
def get_amplitude_phase(real, imag):
    # real = f.real
    # imag = f.imag
    amplitude = np.sqrt(real**2 + imag**2)
    phase = 180/np.pi*np.arctan2(imag, real)
    return amplitude, phase
```

```python
def plot_fields(
    field_plot, freq_ind=0, xlim=500*np.r_[-1, 1], zlim=np.r_[-1000, 10], component="real", ax=None, vmin=None, vmax=None, aspect=True,
    plot_source=False
):
    if ax is None: 
        fig, ax = plt.subplots(1,1)

    if len(field_plot) == mesh.n_faces: 
        field_plot = mesh.average_face_to_cell_vector * np.squeeze(field_plot)
        
        out = mesh.plot_image(
            getattr(field_plot, component), "CCv", view="vec", 
            range_x = xlim, range_y = zlim,
            sample_grid = np.r_[np.diff(xlim)/100, np.diff(zlim)/100],
            mirror=True, #mirror_data=field_mirror,
            ax=ax, 
            pcolor_opts={"norm":LogNorm(vmin=vmin, vmax=vmax)},
            stream_threshold=vmin
        )

    elif len(field_plot) == mesh.n_edges: 
        field_plot = mesh.average_edge_to_cell_vector * np.squeeze(field_plot)
        
        out = mesh.plot_image(
            getattr(field_plot, component), "CC",  
            range_x = xlim, range_y = zlim,
            mirror=True, mirror_data=-1*field_plot,
            ax=ax, 
            pcolor_opts={
                "norm":SymLogNorm(linthresh=vmin if vmin is not None else 1e-12, vmin=-1*vmax if vmax is not None else None, vmax=vmax),
                "cmap":"RdBu_r",
            },
            stream_threshold=vmin
        )

    if plot_source is True: 
        ax.plot(src_locations[:, 0], src_locations[:, 2], "w", lw=2)
    
    cb = plt.colorbar(out[0], ax=ax)
    # cb.set_label("Electric Field (V/m)")
    if aspect is True:
        ax.set_aspect(1)

    return out, cb
```

```python
def getAmmr(mur):
    vol = mesh.cell_volumes
    Div = discretize.utils.sdiag(vol) * mesh.face_divergence
    MeMuI = mesh.get_edge_inner_product(mur * mu_0, invert_matrix=True)
    mui = 1./(mur*mu_0)
    return (
        mesh.edge_curl
        * MeMuI
        * mesh.edge_curl.T.tocsr()
        - Div.T.tocsr()
        * discretize.utils.sdiag(1.0 / vol * mui)
        * Div  # stabalizing term. See (Chen, Haber & Oldenburg 2002)
    )
```

```python
js_dc = line_through_faces(
    mesh, dc_src.location, normalize_by_area=True
)
```

```python
def solve_mmr(mur, dc_fields):
    Ammr = getAmmr(mur)
    Ammr_inv = Pardiso(Ammr)
    rhs = js_dc + dc_fields[:, "j"].flatten()
    mmr_a = Ammr_inv * rhs
    
    MeMuI = mesh.get_edge_inner_product(mur * mu_0, invert_matrix=True)
    b_mmr = mesh.edge_curl.T * mmr_a
    h_mmr = MeMuI * b_mmr

    return {
        "a": mmr_a, 
        "b": b_mmr, 
        "h": h_mmr
    }


    
```
