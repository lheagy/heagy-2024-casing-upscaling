```python
import torch
```

```python
# ----------------------------
# Physical constants (SI)
# ----------------------------
mu_0 = 4.0 * torch.pi * 1e-7
epsilon_0 = 8.8541878128e-12
```

```python
def admittivity(omega, sigma, eps=0.0):
    # sigma + i*omega*eps
    omega = torch.as_tensor(omega)
    sigma = torch.as_tensor(sigma, device=omega.device)
    eps = torch.as_tensor(eps, device=omega.device)
    return sigma + 1j * omega * eps


def gamma(omega, mu, sigma, eps=0.0):
    # γ = sqrt(i ω μ (σ + i ω ε))
    omega = torch.as_tensor(omega)
    mu = torch.as_tensor(mu, device=omega.device)
    sigma = torch.as_tensor(sigma, device=omega.device)
    eps = torch.as_tensor(eps, device=omega.device)
    return torch.sqrt(1j * omega * mu * admittivity(omega, sigma, eps))


def eta_tm(omega, mu, sigma, eps=0.0):
    # η = i ω μ / γ
    g = gamma(omega, mu, sigma, eps)
    return 1j * omega * mu / g
```
